```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    'Dias' : ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta'],
    'Venda' : [500, 200, 1000, 150, 1950]})
```

```python
# Criando o Gráfico
# plt.figure(figsize=(10,6)) define o tamanho da imagem
plt.plot (df['Dias'], df['Venda'], marker='o', linestyle= '-', color='blue')
plt.title ('Vendas por Dias')
plt.xlabel ('Dias da Semana')
plt.ylabel ('Valor Vendido')
plt.show()
```

```python
plt.bar(df['Dias'], df['Venda'], color='blue')
plt.title ('Vendas por Dias')
plt.xlabel ('Dias da Semana')
plt.ylabel ('Valor Vendido')
plt.show()
```

```python
plt.barh(df['Dias'], df['Venda'], color='blue')
plt.title ('Vendas por Dias')
plt.xlabel ('Dias da Semana')
plt.ylabel ('Valor Vendido')
plt.show()
```
