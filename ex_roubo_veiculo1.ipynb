Na última década, algum município da região metropolitana pode ser considerado outlier sobre as ocorrências de furto de veículos?

```python
import pandas as pd
import numpy as np
```

```python
# Usamos o separador de ponto e vírgula (sep=';') e o formato de acentuação correto
dados = pd.read_csv("https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv", encoding='ISO-8859-1', sep=';')
```

```python
# 2. Somar o total de Furtos de Veículo por MUNICÍPIO
# O comando .groupby() junta todos os registros de um mesmo município e o .sum() soma os crimes
resumo_furtos = dados.groupby('munic')['furto_veiculos'].sum().reset_index()
```

```python
# Renomear as colunas para facilitar a leitura
resumo_furtos.columns = ['municipio', 'total_de_furtos']
```

```python
# 3. Encontrar o município com o MAIOR número de furtos (o Outlier)
# Ordenamos a tabela do maior para o menor total...
tabela_ordenada = resumo_furtos.sort_values(by='total_de_furtos',ascending=False)
```

```python
# ... e pegamos o primeiro (iloc[0])
municipio_outlier = tabela_ordenada.iloc[0]
```

```python
# 4. Mostrar o resultado
print("--- Município Outlier em Furto de Veículos (Região Metropolitana) ---")
print(f"Município que mais destoa: {municipio_outlier['municipio']}")
print(f"Total Histórico Acumulado: {municipio_outlier['total_de_furtos']:,.0f}")
```
