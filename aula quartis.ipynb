```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('pesquisa_satisfacao.csv', sep=';')
df
```

```python
media_satisfacao = df.groupby ('Satisfação') ['Tempo de Espera (min)'].mean().reset_index()
media_satisfacao['Tempo de Espera (min)'] = round (media_satisfacao['Tempo de Espera (min)'], 2)
media_satisfacao
```

```python
df_anp = pd.read_csv('ANP2025.csv', sep=';')
df_anp
```

```python
df_anp_rj = df_anp.loc[df_anp['Estado - Sigla'] == 'RJ']
df_anp_rj.head(5)
```

```python
df_anp_rj['Valor de Venda'] = df_anp_rj['Valor de Venda'].str.replace(',', '.').astype(float)
```

```python
media_munc_rj = df_anp_rj.groupby('Municipio') ['Valor de Venda'].mean().reset_index()
media_munc_rj = media_munc_rj.sort_values (by = 'Valor de Venda', ascending = False)
media_munc_rj.head (5)
```

```python
arry_valor_venda = np.array (media_munc_rj['Valor de Venda'])
```

```python
media = np.mean (arry_valor_venda)
mediana = np.median (arry_valor_venda)

q1 = np.quantile (arry_valor_venda, 0.25)
q2 = np.quantile (arry_valor_venda, 0.50)
q3 = np.quantile (arry_valor_venda, 0.75)

print(f'A média é {media}')
print (f'A mediana é {mediana}')
print (f'Primeiro quartil é (Q1){q1}')
print (f'Segundo quartil é (Q2){q2}')
print (f'Terceiro quartil é (Q3){q3}')
```

```python
munic_acima_q3 = media_munc_rj.loc[media_munc_rj['Valor de Venda'] > q3]
munic_acima_q3
```
