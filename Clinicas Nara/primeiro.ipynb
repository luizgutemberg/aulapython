```python
#importar a biblioteca
import pandas as pd
import numpy as np
```

```python
#1: CSV do exercício
consultas = pd.read_csv("consultas.csv")
avaliacoes = pd.read_csv("avaliacoes.csv")
```

```python
#2:  Transformar as colunas de data para o tipo correto - "datatime"
consultas['data_hora_agendada'] = pd.to_datetime(consultas['data_hora_agendada'])
consultas['data_hora_inicio'] = pd.to_datetime(consultas['data_hora_inicio'])
```

```python
#3: Pegar só as consultas que foram relizadas. Não faz sentido realziar analise com consultas que não foram realizadas
consultas_realizadas = consultas[consultas['status'] == 'Realizada'].copy()
```

```python
#4: Calcular o tempo de espera em minutos
# Subtrai a hora agendada da hora que começou e transforma em minutos
consultas_realizadas['tempo_espera_minutos'] = (
    consultas_realizadas['data_hora_inicio'] - consultas_realizadas['data_hora_agendada']
).dt.total_seconds() / 60
```

```python
#5: Criar os quartis Q1, Q2/mediana e  Q3 ---  Q2 é a mediana
print("\n=== Calculo dos Quartis ===")
Q1 = consultas_realizadas['tempo_espera_minutos'].quantile(0.25)
Q2 = consultas_realizadas['tempo_espera_minutos'].quantile(0.50)  # isso é a mediana
Q3 = consultas_realizadas['tempo_espera_minutos'].quantile(0.75)

print(f"Q1 (25% dos dados): {Q1} minutos")
print(f"Q2 (Mediana): {Q2} minutos")
print(f"Q3 (75% dos dados): {Q3} minutos")
```

```python
#6: Identificar os outliers (valores muito estranhos)
# Calcuclo do IQR
print("\n=== PROCURANDO OUTLIERS ===")
IQR = Q3 - Q1
limite_baixo = Q1 - 1.5 * IQR
limite_alto = Q3 + 1.5 * IQR

outliers = consultas_realizadas.loc[
    #(consultas_realizadas['tempo_espera_minutos'] < limite_baixo) |
    consultas_realizadas['tempo_espera_minutos'] > limite_alto
]

print(f'o valor do limite inferior é: {limite_baixo}')
print(f'o valor do limite superior é: {limite_alto}')

outliers
```
