Quais Upps destoam das demais no que diz respeito a ocorrências relacionadas à apreensão de drogas ? 
Usando cálculo estatístico do outlier.

```python
import pandas as pd
import numpy as np
```

```python
dados = pd.read_csv('https://www.ispdados.rj.gov.br/Arquivos/UppEvolucaoMensalDeTitulos.csv', encoding='ISO-8859-1', sep=';')
dados.columns
```

```python
# 2. Somar o total de apreensão de drogas por CISP (Delegacia)
# Agrupamos todas as ocorrências de apreensão_drogas por cada código de delegacia.
resumo_drogas = dados.groupby('upp')['apreensao_drogas'].sum().reset_index()
resumo_drogas.columns = ['upp', 'total_apreensao']
```

```python
# 3. Calcular o Outlier usando o Intervalo Interquartil (IQR)
# O IQR define um limite estatístico para o que é considerado 'normal'.

# Calcular o 25º Percentil (Q1) e o 75º Percentil (Q3)
Q1 = np.percentile(resumo_drogas['total_apreensao'], 25)
Q3 = np.percentile(resumo_drogas['total_apreensao'], 75)
```

```python
# Calcular o Intervalo Interquartil (IQR)
IQR = Q3 - Q1
```

```python
# O Limite Superior (qualquer valor acima disso é um outlier)
limite_superior = Q3 + 1.5 * IQR
```

```python
# 4. Identificar a(s) CISP(s) que são Outliers
# Filtramos a tabela para mostrar apenas as CISPs acima do limite
outliers = resumo_drogas[resumo_drogas['total_apreensao'] > limite_superior]
```

```python
#Encontrar o Outlier Mais Extremo (o que tem o maior total)
outlier_extremo = outliers.sort_values(by='total_apreensao', ascending=False).iloc[0]
```

```python
# 5. Mostrar o Resultado
print("--- Análise Estatística de Apreensão de Drogas por UPP ---")
print(f"Q1 (Limite Inferior da 'Normalidade'): {Q1:,.0f}")
print(f"Limite Estatístico Superior (Outlier): {limite_superior:,.0f}")
print("----------------------------------------------------------")

print("\n--- UPP que mais destoa (Outlier Extremo) ---")
print(f"UPP Outlier: {outlier_extremo['upp']}")
print(f"Total Histórico Acumulado: {outlier_extremo['total_apreensao']:,.0f}")
print("----------------------------------------------------------")
```
